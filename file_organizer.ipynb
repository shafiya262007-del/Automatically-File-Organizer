```python
import os
import shutil
import tkinter as tk
from tkinter import filedialog, messagebox

# Create window
root = tk.Tk()
root.title("Automated File Organizer")
root.geometry("500x400")
root.configure(bg="#EAF2F8")
root.resizable(False, False)

selected_folder = ""

# Select folder
def select_folder():
    global selected_folder

    selected_folder = filedialog.askdirectory()

    if selected_folder:
        folder_label.config(
            text=selected_folder
        )

# Organize files
def organize_files():

    if not selected_folder:
        messagebox.showwarning(
            "Warning",
            "Please select a folder first!"
        )
        return

    categories = {
        "Images": [".jpg", ".jpeg", ".png", ".gif"],
        "Documents": [".pdf", ".docx", ".txt", ".doc"],
        "Videos": [".mp4", ".mkv", ".avi"],
        "Audio": [".mp3", ".wav"],
        "Others": []
    }

    moved = 0

    try:
        for filename in os.listdir(selected_folder):

            source = os.path.join(
                selected_folder, filename
            )

            if not os.path.isfile(source):
                continue

            extension = os.path.splitext(
                filename
            )[1].lower()

            category = "Others"

            for folder, extensions in categories.items():
                if extension in extensions:
                    category = folder
                    break

            destination_folder = os.path.join(
                selected_folder, category
            )

            os.makedirs(
                destination_folder,
                exist_ok=True
            )

            destination = os.path.join(
                destination_folder, filename
            )

            # Avoid overwriting existing files
            if os.path.exists(destination):
                name, ext = os.path.splitext(filename)
                count = 1

                while os.path.exists(destination):
                    new_name = f"{name}_{count}{ext}"
                    destination = os.path.join(
                        destination_folder, new_name
                    )
                    count += 1

            shutil.move(source, destination)
            moved += 1

        result_label.config(
            text=f"Successfully organized {moved} files!"
        )

        messagebox.showinfo(
            "Success",
            f"{moved} files organized successfully!"
        )

    except OSError as error:
        messagebox.showerror(
            "Error",
            str(error)
        )

# Title
title = tk.Label(
    root,
    text="AUTOMATED FILE ORGANIZER",
    font=("Arial", 19, "bold"),
    bg="#1F4E79",
    fg="white",
    pady=20
)
title.pack(fill="x")

# Main frame
frame = tk.Frame(
    root,
    bg="white",
    padx=25,
    pady=25
)
frame.pack(padx=25, pady=25, fill="both", expand=True)

tk.Label(
    frame,
    text="Select a folder to organize",
    font=("Arial", 13, "bold"),
    bg="white"
).pack(pady=10)

# Select button
tk.Button(
    frame,
    text="SELECT FOLDER",
    command=select_folder,
    bg="#3498DB",
    fg="white",
    font=("Arial", 11, "bold"),
    padx=15,
    pady=8
).pack(pady=10)

# Folder path
folder_label = tk.Label(
    frame,
    text="No folder selected",
    bg="white",
    fg="gray",
    wraplength=350
)
folder_label.pack(pady=10)

# Organize button
tk.Button(
    frame,
    text="ORGANIZE FILES",
    command=organize_files,
    bg="#27AE60",
    fg="white",
    font=("Arial", 12, "bold"),
    padx=15,
    pady=10
).pack(pady=15)

# Result
result_label = tk.Label(
    frame,
    text="Result will appear here",
    bg="white",
    fg="#16803C",
    font=("Arial", 12, "bold")
)
result_label.pack(pady=10)

root.mainloop()
```
